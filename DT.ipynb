```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
x = dataset.values[:,:len(dataset.columns)-1]
print(x.shape)
y = dataset.values[:,len(dataset.columns)-1]
print(y.shape)
# # 334 is for dt
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size = 0.20, random_state = 334)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
scaler.fit(X_train)
scale_x_train=scaler.transform(X_train)
print(scale_x_train.shape)
scale_x_test=scaler.transform(X_test)
```

```python
dtc = DecisionTreeClassifier(criterion='gini', max_depth=5, random_state=0)
dtc.fit(scale_x_train, Y_train)
train_accuracy = dtc.score(scale_x_train, Y_train)
test_accuracy = dtc.score(scale_x_test, Y_test)
print("Train-Acc:",train_accuracy*100)
print("Test-Acc:",test_accuracy*100)
print(abs(train_accuracy-test_accuracy))
```
